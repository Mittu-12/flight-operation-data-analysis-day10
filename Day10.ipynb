This notebook follows a complete data analysis workflow on the provided Flight Operations dataset using Pandas: loading and understanding the data, inspecting its structure and statistics, selecting and filtering relevant records, sorting, grouping and aggregation, applying transformations, and concluding with key observations.

```python
import pandas as pd
import numpy as np
print('Pandas version:', pd.__version__)
```

## 1. Loading and Understanding the Dataset

```python
df = pd.read_csv('Day10_Flight_Operations_Dataset.csv')
print("Shape:", df.shape)
df.head()
```

```python
df.columns.tolist()
```

## 2. Inspecting Structure and Statistics

```python
df.info()
```

```python
# Checking for missing values
df.isna().sum()
```

```python
# Numerical summary statistics
df.describe()
```

```python
# Categorical overview
for col in ['Airline', 'Travel_Class', 'Flight_Status', 'Weather', 'Booking_Channel']:
    print(f"\n{col}:")
    print(df[col].value_counts())
```

## 3. Handling Missing Values
`Delay_Minutes` has some missing values — these correspond to cancelled flights, which never departed and therefore have no delay recorded.

```python
missing_delay = df[df['Delay_Minutes'].isna()]
print("Flights with missing Delay_Minutes:", missing_delay.shape[0])
missing_delay['Flight_Status'].value_counts()
```

```python
# Fill missing delay minutes with 0 for cancelled flights (no delay applicable)
df['Delay_Minutes'] = df['Delay_Minutes'].fillna(0)
print("Remaining missing values:", df['Delay_Minutes'].isna().sum())
```

## 4. Data Selection

```python
# Selecting key columns for a quick operations view
ops_view = df[['Flight_ID', 'Airline', 'Origin', 'Destination', 'Flight_Status', 'Delay_Minutes']]
ops_view.head()
```

## 5. Filtering

```python
# Flights delayed by more than 30 minutes
long_delays = df[df['Delay_Minutes'] > 30]
print("Flights delayed more than 30 minutes:", long_delays.shape[0])
long_delays[['Flight_ID', 'Airline', 'Origin', 'Destination', 'Delay_Minutes']].head()
```

```python
# Business class flights with high passenger satisfaction (>=4)
happy_business = df[(df['Travel_Class'] == 'Business') & (df['Passenger_Satisfaction'] >= 4)]
print("Satisfied Business class flights:", happy_business.shape[0])
happy_business[['Flight_ID', 'Airline', 'Passenger_Satisfaction']].head()
```

```python
# Cancelled flights during storms
storm_cancellations = df[(df['Flight_Status'] == 'Cancelled') & (df['Weather'] == 'Storm')]
storm_cancellations[['Flight_ID', 'Airline', 'Origin', 'Destination', 'Weather']]
```

## 6. Sorting

```python
# Top 5 most delayed flights
most_delayed = df.sort_values('Delay_Minutes', ascending=False).head(5)
most_delayed[['Flight_ID', 'Airline', 'Origin', 'Destination', 'Delay_Minutes']]
```

```python
# Flights sorted by passenger load (Passengers / Seat_Capacity), highest first
df['Load_Factor'] = (df['Passengers'] / df['Seat_Capacity']).round(3)
fullest_flights = df.sort_values('Load_Factor', ascending=False).head(5)
fullest_flights[['Flight_ID', 'Airline', 'Passengers', 'Seat_Capacity', 'Load_Factor']]
```

## 7. Grouping and Aggregation

```python
# Average delay and flight status counts by airline
airline_summary = df.groupby('Airline').agg(
    Total_Flights=('Flight_ID', 'count'),
    Avg_Delay_Minutes=('Delay_Minutes', 'mean'),
    Avg_Satisfaction=('Passenger_Satisfaction', 'mean'),
    Avg_Ticket_Price=('Average_Ticket_Price', 'mean')
).sort_values('Avg_Delay_Minutes', ascending=False)

airline_summary.round(2)
```

```python
# Flight status breakdown by airline
status_by_airline = df.groupby(['Airline', 'Flight_Status']).size().unstack(fill_value=0)
status_by_airline
```

```python
# Average delay and satisfaction by weather condition
weather_summary = df.groupby('Weather').agg(
    Total_Flights=('Flight_ID', 'count'),
    Avg_Delay_Minutes=('Delay_Minutes', 'mean'),
    Avg_Satisfaction=('Passenger_Satisfaction', 'mean')
).sort_values('Avg_Delay_Minutes', ascending=False)

weather_summary.round(2)
```

```python
# Average ticket price and satisfaction by travel class
class_summary = df.groupby('Travel_Class').agg(
    Total_Flights=('Flight_ID', 'count'),
    Avg_Ticket_Price=('Average_Ticket_Price', 'mean'),
    Avg_Satisfaction=('Passenger_Satisfaction', 'mean'),
    Avg_Baggage_Kg=('Avg_Baggage_Kg', 'mean')
)
class_summary.round(2)
```

```python
# Busiest routes (Origin-Destination pairs)
df['Route'] = df['Origin'] + ' -> ' + df['Destination']
route_summary = df.groupby('Route').agg(
    Total_Flights=('Flight_ID', 'count'),
    Avg_Delay_Minutes=('Delay_Minutes', 'mean'),
    Total_Passengers=('Passengers', 'sum')
).sort_values('Total_Flights', ascending=False)

route_summary.head(10).round(2)
```

```python
# Booking channel popularity and average satisfaction
channel_summary = df.groupby('Booking_Channel').agg(
    Total_Bookings=('Flight_ID', 'count'),
    Avg_Satisfaction=('Passenger_Satisfaction', 'mean')
).sort_values('Total_Bookings', ascending=False)

channel_summary.round(2)
```

## 8. Data Transformations

```python
# Categorize delay severity
def delay_category(minutes):
    if minutes == 0:
        return 'On Time'
    elif minutes <= 15:
        return 'Minor Delay'
    elif minutes <= 45:
        return 'Moderate Delay'
    else:
        return 'Severe Delay'

df['Delay_Category'] = df['Delay_Minutes'].apply(delay_category)
df['Delay_Category'].value_counts()
```

```python
# Extract month and day-of-week from Flight_Date
df['Flight_Date'] = pd.to_datetime(df['Flight_Date'])
df['Flight_Month'] = df['Flight_Date'].dt.month_name()
df['Flight_Weekday'] = df['Flight_Date'].dt.day_name()

df[['Flight_ID', 'Flight_Date', 'Flight_Month', 'Flight_Weekday']].head()
```

```python
# Revenue estimate per flight (Passengers * Average_Ticket_Price)
df['Estimated_Revenue'] = df['Passengers'] * df['Average_Ticket_Price']
df[['Flight_ID', 'Airline', 'Passengers', 'Average_Ticket_Price', 'Estimated_Revenue']].sort_values('Estimated_Revenue', ascending=False).head(5)
```

## 9. Answering Key Questions

```python
# Which airline has the highest on-time performance rate?
on_time_rate = df.groupby('Airline').apply(
    lambda x: (x['Flight_Status'] == 'On Time').mean() * 100, include_groups=False
).sort_values(ascending=False)
print("On-time performance rate by airline (%):")
on_time_rate.round(1)
```

```python
# Which weekday sees the most delays on average?
weekday_delay = df.groupby('Flight_Weekday')['Delay_Minutes'].mean().sort_values(ascending=False)
print("Average delay by day of week (minutes):")
weekday_delay.round(2)
```

```python
# Does higher ticket price correlate with higher satisfaction?
price_satisfaction_corr = df['Average_Ticket_Price'].corr(df['Passenger_Satisfaction'])
print(f"Correlation between ticket price and passenger satisfaction: {price_satisfaction_corr:.3f}")
```

## 10. Key Observations

1. **Cancellations are weather-driven** — all cancelled flights in the dataset occurred during storm conditions, suggesting weather is the primary disruption factor rather than airline operational issues.
2. **Delay severity varies notably by airline** — some airlines show consistently higher average delay minutes than others, which could reflect differences in scheduling buffers or route difficulty.
3. **Storms and fog cause the longest average delays**, while flights in clear weather are delayed far less on average — reinforcing weather as a major operational risk factor.
4. **Business class passengers report higher average satisfaction** than Economy passengers, likely tied to service quality and legroom rather than delay performance alone.
5. **A small number of routes account for a large share of total flights**, indicating operations are concentrated on a few high-traffic corridors.
6. **Ticket price shows only a weak correlation with passenger satisfaction**, suggesting price alone doesn't drive how satisfied passengers feel — service and punctuality likely matter more.
7. **On-time performance rates differ meaningfully across airlines**, which could be a useful metric for passengers choosing carriers on delay-sensitive routes.
8. **Load factors vary widely across flights**, with some routes running near full capacity while others are comparatively underbooked — useful for capacity planning.
